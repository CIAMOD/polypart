```python
from polypart.experiments import (
    ArrangementClass,
    Experiment,
    PolytopeClass,
    plot_experiment_summary,
    print_results_summary,
    run_experiments,
)
from polypart.experiments.report import (
    build_results_table,
    plot_moduli_report,
    plot_permutahedron_report,
    plot_random_report,
    results_to_latex,
)
```

## Moduli Experiments


```python
experiments_n1 = [
    Experiment(
        PolytopeClass("moduli_n1"),
        ArrangementClass("moduli_n1"),
        d=d,
    )
    for d in [2, 3, 4, 5, 6, 7]
]
experiments_r2 = [
    Experiment(
        PolytopeClass("moduli_r2"),
        ArrangementClass("moduli_r2"),
        d=d,
    )
    for d in [2, 3, 4, 5, 6]
]

plot_moduli_report(experiments_n1, experiments_r2, folder="../data/cluster")
```

## Permutahedron Arrangement (Affine Braid)


```python
experiments = [
    Experiment(
        PolytopeClass("cube", bbox_mode=True),
        ArrangementClass("perm"),
        d=d,
    )
    for d in [3, 4, 5, 6, 7, 8, 9]
]

plot_permutahedron_report(experiments, folder="../data/cluster")
```

```python
tbl = build_results_table(experiments, folder="/home/jovyan/work/polypart/data/cluster")
latex_code = results_to_latex(tbl)
with open("./work/results_table.tex", "w") as f:
    f.write(latex_code)
```
