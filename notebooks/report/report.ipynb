```python
from polypart.experiments import (
    ArrangementClass,
    Experiment,
    PolytopeClass,
    plot_experiment_summary,
    print_results_summary,
    run_experiments,
)
from polypart.experiments.report import (
    build_results_table,
    plot_experiment_summary_paper,
    plot_moduli_n1_report,
    plot_moduli_r2_report,
    plot_permutahedron_report,
    plot_random_report,
    results_to_latex,
)

DATA_FOLDER = "../../data/clusterv2/"
```

## Moduli Experiments


```python
experiments_n1 = [
    Experiment(
        PolytopeClass("moduli_n1"),
        ArrangementClass("moduli_n1"),
        d=d,
    )
    for d in [2, 3, 4, 5, 6, 7, 8]
]

plot_moduli_n1_report(experiments_n1, folder=DATA_FOLDER + "/moduli_n1/")
print_results_summary(experiments_n1[-2], folder=DATA_FOLDER + "/moduli_n1/")
plot_experiment_summary(experiments_n1[-2], folder=DATA_FOLDER + "/moduli_n1/")
plot_experiment_summary_paper(experiments_n1[-2], folder=DATA_FOLDER + "/moduli_n1/")
```

```python
experiments_n1_rng = [
    Experiment(
        PolytopeClass("moduli_n1"),
        ArrangementClass("random", m=m, degen_ratio=0.0),
        d=d,
    )
    for d, m in zip([2, 3, 4, 5, 6, 7], [1, 3, 11, 21, 65, 129])
]

plot_moduli_n1_report(experiments_n1_rng, folder=DATA_FOLDER + "/moduli_n1_random/")
plot_experiment_summary(
    experiments_n1_rng[-1], folder=DATA_FOLDER + "/moduli_n1_random/"
)
print_results_summary(experiments_n1_rng[-1], folder=DATA_FOLDER + "/moduli_n1_random/")
plot_experiment_summary_paper(
    experiments_n1_rng[-1],
    folder=DATA_FOLDER + "/moduli_n1_random/",
    remove_titles=True,
)
```

```python
experiments_r2 = [
    Experiment(
        PolytopeClass("moduli_r2"),
        ArrangementClass("moduli_r2"),
        d=d,
    )
    for d in [2, 3, 4, 5, 6, 7]
]

plot_moduli_r2_report(experiments_r2, folder=DATA_FOLDER + "/moduli_r2/")
```
